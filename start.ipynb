## 1. Data Cleaning : 

### A. Removing The Useless Columns :

```python
df.head(0)
```

```python
df = df.drop(columns=['Poster_Link', 'Overview'])
```

```python
df.head(4)
```

### B. Replace NULL values with Passed in Certificate column :

```python
# In this dataset there are muliple certificates of movies are not available.
# so i m assuming that these movies are passed.


df['Certificate'] = df['Certificate'].replace(np.nan, 'Passed')
```

### C. Replacing the NULL values in Meta score :

```python
df['Meta_score'].isnull().sum()
```

```python
# there are around 157 values are null in this column 
# this data is important for me thats why i m using ffill for replacing null values. 


df['Meta_score']= df['Meta_score'].ffill()
```
